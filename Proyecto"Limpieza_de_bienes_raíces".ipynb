#### Ejercicio 18. Normaliza la variable de precios para cada población y traza los 4 histogramas en el mismo gráfico (debes usar el subconjunto obtenido en la pregunta 14) (★★★)

Para el método de normalización, puedes usar el que consideres adecuado, no hay una única respuesta correcta para esta pregunta. Imprime el gráfico y escribe en la celda de Markdown un breve análisis sobre el gráfico.

Pista: Puedes ayudarte revisando la demostración multihist de Matplotlib.

```python
df_cinturon_sur_clean = df_cinturon_sur.dropna(subset= ["price"])
df_cinturon_sur_clean["price_normalized"] = df_cinturon_sur_clean.groupby("level5")["price"].transform(lambda x: (x - x.mean())/ x.std())
poblaciones_unicas = df_cinturon_sur_clean["level5"].unique()

plt.figure(figsize=(12,7))
for poblacion in poblaciones_unicas:
  precios_normal_poblacion = df_cinturon_sur_clean[df_cinturon_sur_clean["level5"]== poblacion]["price_normalized"]
  plt.hist(precios_normal_poblacion, bins=30, alpha = 0.75, label=poblacion)
plt.show()
```

Como se puede observar en los histogramas, las viviendas en las poblaciones del cinturón sur muestran un claro sesgo a la derecha. Esto nos indica que la mayoría de las viviendas tienen precios concentrados alrededor de la media de cada población, mientras que existen algunas propiedades con precios significativamente más altos que la media, extendiendo la 'cola' de la distribución hacia la derecha. Esta observación es consistente con la existencia de viviendas de lujo o ubicadas en zonas de alto valor, lo cual concuerda con las observaciones y gráficos anteriormente realizados (por ejemplo, los outliers en el scatter plot del Ejercicio 11)

#### Ejercicio 21. Realiza un trazado de las coordenadas (columnas latitud y longitud) del cinturón sur de Madrid por color de cada población (debes usar el subconjunto obtenido del Ejercicio 14) (★★★★)

Ejecuta la siguiente celda y luego comienza a codear en la siguiente. Debes implementar un código simple que transforme las columnas de coordenadas en un diccionario de Python (agrega más información si es necesario) y agrégala al mapa.

```python
from ipyleaflet import Map, basemaps

# Mapa centrado en Madrid
map = Map(center = (40.35, -3.8), zoom = 11.5, min_zoom = 1, max_zoom = 20,
    basemap=basemaps.OpenStreetMap.Mapnik)
map
```

```python
from ipyleaflet import Map, basemaps, CircleMarker, LayerGroup, LayersControl
import pandas as pd
import numpy as np

df_map = df_cinturon_sur.dropna(subset=["latitude", "longitude", "level5"]).copy()

poblaciones_unicas = df_map["level5"].unique()
colores_por_poblacion = {
    'Alcorcón': "red",
    'Fuenlabrada': "blue",
    'Getafe': "green",
    'Leganés': "purple",
}


for poblacion in poblaciones_unicas:
    df_poblacion = df_map[df_map["level5"] == poblacion]
    markers = []
    for idx, row in df_poblacion.iterrows():
        marker = CircleMarker(
            location=(row["latitude"], row["longitude"]),
            radius=3,
            color=colores_por_poblacion.get(poblacion, "gray"),
            fill_color=colores_por_poblacion.get(poblacion, "gray"),
            fill_opacity=0.7,
            weight=1
        )
        markers.append(marker)
    poblacion_layer = LayerGroup(layers=tuple(markers), name=poblacion)
    map.add_layer(poblacion_layer)

map
```
